```python
import pandas as pd
ser=pd.Series(['123','233.$','400.$'],index=[1,3,4])
ser
```

```python
def removed(x):
    if x.endswith('$'):
        return int(x[:-2])
    else:
        return x

si=ser.map(removed)
si
```
