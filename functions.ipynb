wap using the func to generate perfect numbers from 1 to 1000

```python
def is_perfect(n):
    sum = 0

    for i in range(1, n):
        if n % i == 0:
            sum += i

    return sum == n


print("Perfect numbers from 1 to 1000:")

for num in range(1, 1001):
    if is_perfect(num):
        print(num)
```

wap to convert the temprature into either celciuc or farenheite user provides the temp along its unit

```python
def tempcon(temp,unit):
    if unit in "Cc":
        return temp * 9/5 + 32
    elif unit in "Ff":
        return (temp - 32) * 5/9
    else:
        print("Invalid unit. Please enter 'C' for Celsius or 'F' for Fahrenheit.")

t=int(input("Enter the temperature: "))
u=input("Enter the unit (C or F): ")
print(tempcon(t,u)) 
```

```python
# password strenth cheker
def check_password_strength(password):
    if len(password) < 8:
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char in "!@#$%^&*()-_=+" for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    return True

passw = input("Enter a password: ")
if check_password_strength(passw):
    print("Password is strong.")
else:
    print("Password is weak.")
```

```python
# wap using function to calculate the total cost of items in a shopping cart
def calculate_total_cost(cart):
    total_cost = 0
    for item in cart:
        total_cost += item['price'] * item['quantity']
    return total_cost

# Example usage:    
cart=[
    {'name': 'bottle', 'price': 10, 'quantity': 2},
    {'name': 'shirt', 'price': 5, 'quantity': 3},
    {'name': 'perfume', 'price': 8, 'quantity': 1}
]
total = calculate_total_cost(cart)
print(f"Total cost: ${total}")
```

```python
# wap to find the factorial of a given number using recursion
def fac(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * fac(n - 1) 
    
print(fac(5))
print(fac(0))
print(fac(1))     

```

```python
# map() map function applies a given function to all items in an iterable (like a list) and returns a map object (which is an iterator). You can convert the map object to a list or other iterable types.
# Example usage:
numbers = [1, 2, 3, 4, 5]
def square(x):
    return x ** 2

print(square(2))
print (map(lambda x: x**2, numbers))
print(list(map(lambda x: x**2, numbers)))  
```
