```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
iris=pd.read_csv("iris_dataset.csv")
print(type(iris))
print(iris.shape)
print(iris.head())
```

```python
semesters=np.arange(1,7)
cgpa=[7.1,8.5,8,9.0,9.2,9.3]
fig,ax=plt.subplots(figsize=(6,3.5))
ax.plot(semesters,cgpa,marker='o',color='Fuchsia',linestyle='dashed')
ax.set_xticks(semesters)
ax.set_yticks(np.arange(6.5,9.1,0.5))
ax.grid(axis='x',linestyle='dotted',alpha=0.61)
ax.set(title="CGPA Trend",xlabel="Semesters",ylabel="CGPA")
plt.show()
```
